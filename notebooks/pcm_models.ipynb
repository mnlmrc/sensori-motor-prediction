```python
import pickle
import os
import matplotlib.pyplot as plt
import numpy as np
import SensoriMotorPrediction.globals as gl
import SensoriMotorPrediction.pcm_models as pcm_models
import PcmPy as pcm
```

```python
M_exec = pcm_models.make_execution_models()
M_exec = M_exec[1:]

models = ['finger', 'expectation', 'surprise']

fig, axs = plt.subplots(1, len(models), sharex=True, sharey=True)

for m, model in enumerate(models):
    ax = axs[m]
    G = (M_exec[m].G)
    D = pcm.G_to_dist(G)
    ax.imshow(D)
    ax.axhline(3.5, lw=.8, color='k')
    ax.axvline(3.5, lw=.8, color='k')
    ax.set_title(model)

axs[0].set_yticks(np.arange(8))
axs[0].set_yticklabels(gl.conditions[5:])
#axs[0].set_xticks([])

fig.savefig('figures/execution_models.pdf')

plt.show()
```
